# 1️⃣ What is an AI Agent?

## Definition

An AI agent is a software system that can understand a goal,
reason about the required steps, use tools when necessary,
process the results, and produce an appropriate response.

Unlike a traditional chatbot that mainly responds to messages,
an AI agent can take actions to accomplish a task.

---

## Chatbot vs AI Agent

| Feature | Chatbot | AI Agent |
|---|---|---|
| Conversation | ✅ | ✅ |
| Understands instructions | ✅ | ✅ |
| Uses external tools | Limited | ✅ |
| Takes actions | Limited | ✅ |
| Multi-step tasks | Limited | ✅ |
| Tool selection | Usually no | ✅ |
| Decision making | Limited | ✅ |
| Memory / State | Optional | Often required |

---

## Basic AI Agent Workflow

User Request
↓
Understand Task
↓
Decide Action
↓
Select Tool
↓
Execute Tool
↓
Process Result
↓
Generate Response

# 2️⃣ Function Calling & Tool Calling

## What is Tool Calling?

Tool calling allows an AI model to request the execution of
a specific function when it needs external information or
an action.

The LLM decides which tool is useful, while the application
executes the actual function.

---

## Why are Tools Important?

An LLM by itself mainly generates responses.

Tools allow an AI agent to:

- Perform calculations
- Search information
- Access databases
- Call APIs
- Create or update records
- Perform application-specific actions

---

## Tool Calling Workflow

| Step | Action |
|---|---|
| 1️⃣ | 👤 User gives a request |
| 2️⃣ | 🧠 LLM analyzes the request |
| 3️⃣ | 🛠️ LLM selects a tool |
| 4️⃣ | ⚙️ Application executes the tool |
| 5️⃣ | 📊 Tool returns a result |
| 6️⃣ | 🤖 LLM uses the result |
| 7️⃣ | 💡 Final response is generated |

---

## Example

User:

"What is 125 × 48?"

The agent can identify that a calculator tool is required.

The calculator performs:

125 × 48 = 6000

The result is then provided to the LLM so it can respond
to the user.

---

## Important Concept

The LLM does **not automatically execute Python functions**.

Instead:

**LLM → Tool Request → Application Executes Tool → Tool Result → LLM → Final Answer**

This separation makes tool-using agents more controllable
and easier to evaluate.

```python
# ============================================
# TOOL 1: CALCULATOR
# ============================================

def calculator(expression):
    """
    Calculator tool for basic mathematical expressions.
    """

    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return f"Calculation result: {result}"

    except Exception:
        return "Unable to calculate the expression."


# Test the calculator
print(calculator("125 * 48"))
print(calculator("250 + 375"))
print(calculator("1000 / 25"))
```

```python
# ============================================
# TOOL 2: TASK MANAGEMENT
# ============================================

tasks = []


def add_task(task):
    tasks.append(task)
    return f"Task added successfully: {task}"


def list_tasks():
    if not tasks:
        return "No tasks found."

    result = "Current tasks:\n"

    for i, task in enumerate(tasks, start=1):
        result += f"{i}. {task}\n"

    return result


def remove_task(task_number):

    try:
        task_number = int(task_number)

        if 1 <= task_number <= len(tasks):
            removed = tasks.pop(task_number - 1)
            return f"Task removed: {removed}"

        return "Invalid task number."

    except Exception:
        return "Invalid task number."


# Test task tools
print(add_task("Complete Module 5"))
print(add_task("Upload project to GitHub"))
print(list_tasks())
```

```python
# ============================================
# STEP 8: AI AGENT WITH TOOL CALLING
# ============================================

tools = [
    {
        "type": "function",
        "function": {
            "name": "calculator",
            "description": "Calculate a mathematical expression.",
            "parameters": {
                "type": "object",
                "properties": {
                    "expression": {
                        "type": "string",
                        "description": "Mathematical expression to calculate."
                    }
                },
                "required": ["expression"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "add_task",
            "description": "Add a new task to the task list.",
            "parameters": {
                "type": "object",
                "properties": {
                    "task": {
                        "type": "string",
                        "description": "The task to add."
                    }
                },
                "required": ["task"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "list_tasks",
            "description": "Show all current tasks.",
            "parameters": {
                "type": "object",
                "properties": {},
                "required": []
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "remove_task",
            "description": "Remove a task using its task number.",
            "parameters": {
                "type": "object",
                "properties": {
                    "task_number": {
                        "type": "integer",
                        "description": "The number of the task to remove."
                    }
                },
                "required": ["task_number"]
            }
        }
    }
]

print("✅ Agent tools defined!")
print("Available tools:")
print("🧮 calculator")
print("➕ add_task")
print("📋 list_tasks")
print("🗑️ remove_task")
```

```python
# ============================================
# STEP 9: AGENT DECISION + TOOL EXECUTION
# ============================================

import json


def execute_tool(tool_name, arguments):

    if tool_name == "calculator":
        return calculator(arguments["expression"])

    elif tool_name == "add_task":
        return add_task(arguments["task"])

    elif tool_name == "list_tasks":
        return list_tasks()

    elif tool_name == "remove_task":
        return remove_task(arguments["task_number"])

    else:
        return "Unknown tool."


def run_agent(user_request):

    messages = [
        {
            "role": "system",
            "content": """
You are an AI task management agent.

Analyze the user's request and use the available tools
when necessary.

Use the calculator for mathematical calculations.

Use task tools when the user wants to add, view, or remove tasks.

If no tool is required, answer normally.
"""
        },
        {
            "role": "user",
            "content": user_request
        }
    ]

    # First LLM call: decide whether a tool is required
    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        tools=tools,
        max_tokens=300
    )

    assistant_message = response.choices[0].message

    # If the model wants to use a tool
    if getattr(assistant_message, "tool_calls", None):

        messages.append(assistant_message)

        for tool_call in assistant_message.tool_calls:

            tool_name = tool_call.function.name

            arguments = json.loads(
                tool_call.function.arguments
            )

            print(f"🛠️ Agent selected tool: {tool_name}")
            print(f"📥 Arguments: {arguments}")

            tool_result = execute_tool(
                tool_name,
                arguments
            )

            print(f"📤 Tool result: {tool_result}")

            messages.append(
                {
                    "role": "tool",
                    "tool_call_id": tool_call.id,
                    "content": str(tool_result)
                }
            )

        # Second LLM call: generate final response
        final_response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            max_tokens=300
        )

        return final_response.choices[0].message.content

    else:

        return assistant_message.content


# Test the AI Agent
result = run_agent(
    "Calculate 125 multiplied by 48"
)

print("\n🤖 FINAL AGENT RESPONSE")
print("=" * 50)
print(result)
```

```python
# ============================================
# STEP 10: TEST MULTIPLE AGENT ACTIONS
# ============================================

test_requests = [
    "Add a task called Complete Module 5",
    "Add a task called Update my GitHub repository",
    "Show me my current tasks",
    "Calculate 250 * 24"
]

for request in test_requests:

    print("\n" + "=" * 60)
    print("👤 USER:", request)
    print("=" * 60)

    result = run_agent(request)

    print("🤖 AGENT:", result)
```

# 3️⃣ Agent Memory & State

## What is Agent State?

Agent state is the information an agent maintains while
working on a task or conversation.

State can include:

- User requests
- Previous messages
- Tool results
- Current tasks
- Intermediate results
- Workflow status

---

## What is Agent Memory?

Memory allows an AI system to retain useful information
between interactions.

For example, if a user says:

"Add Complete Module 5 to my tasks."

The agent stores the task.

If the user later asks:

"Show my tasks."

The agent can retrieve the previously stored task.

---

## Short-Term vs Long-Term Memory

| Memory Type | Description |
|---|---|
| Short-Term Memory | Information from the current interaction |
| Long-Term Memory | Information stored for future interactions |

---

## State in Our Project

Our task list acts as a simple form of agent state.

The variable:

`tasks`

stores the current tasks handled by the agent.

The agent can:

- Add tasks
- Retrieve tasks
- Remove tasks
- Use the current state when responding

---

## Why Memory Matters

Memory helps agents perform multi-step tasks and maintain
context across interactions.

A useful agent should know what has already happened before
deciding what to do next.

---

## Key Takeaway

**Agent + Tools + State = More Capable AI Workflow**

The agent uses tools to perform actions and maintains state
to keep track of the results.

```python
# ============================================
# STEP 12: TRUE MULTI-STEP AI AGENT
# ============================================

import json


def run_multi_step_agent(user_request, max_steps=5):

    messages = [
        {
            "role": "system",
            "content": """
You are an AI task management agent.

You can use tools to complete the user's request.

IMPORTANT RULES:
1. Use tools whenever the request requires an action.
2. Never claim that an action was completed unless the tool
   was actually executed successfully.
3. If multiple actions are required, continue using tools
   until the complete request is finished.
4. After every tool result, decide whether another tool is needed.
5. Only give the final response after completing all required actions.
"""
        },
        {
            "role": "user",
            "content": user_request
        }
    ]

    for step in range(max_steps):

        print(f"\n🔄 Agent Step {step + 1}")

        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            tools=tools,
            tool_choice="auto",
            max_tokens=300
        )

        assistant_message = response.choices[0].message

        # No more tools required
        if not getattr(assistant_message, "tool_calls", None):

            print("💡 No additional tool required.")

            return assistant_message.content

        # Add assistant tool request to conversation
        messages.append(assistant_message)

        # Execute every requested tool
        for tool_call in assistant_message.tool_calls:

            tool_name = tool_call.function.name

            arguments = json.loads(
                tool_call.function.arguments
            )

            print(f"🛠️ Tool selected: {tool_name}")
            print(f"📥 Arguments: {arguments}")

            tool_result = execute_tool(
                tool_name,
                arguments
            )

            print(f"📤 Tool result: {tool_result}")

            # Give tool result back to the LLM
            messages.append(
                {
                    "role": "tool",
                    "tool_call_id": tool_call.id,
                    "content": str(tool_result)
                }
            )

    return "The agent reached the maximum number of tool steps."


# ============================================
# TEST MULTI-STEP TASK
# ============================================

multi_step_request = """
Calculate 75 * 16 and then add a task called
"Review the calculation result".
"""

print("👤 USER REQUEST:")
print(multi_step_request)

final_result = run_multi_step_agent(
    multi_step_request
)

print("\n🤖 FINAL AGENT RESPONSE")
print("=" * 60)
print(final_result)

print("\n📋 CURRENT TASKS")
print("=" * 60)
print(list_tasks())
```

# 5️⃣ AI Agent Architecture & Frameworks

## Agent Architecture

An AI agent generally contains several important components:

| Component | Purpose |
|---|---|
| 🧠 LLM | Understands requests and decides actions |
| 🛠️ Tools | Perform external actions |
| 🧩 Agent Logic | Controls the workflow |
| 💾 State | Stores current information |
| 📝 Memory | Retains useful information |
| 🛡️ Safety | Controls tool usage and permissions |

---

## Our Agent Architecture

Our project uses:

**User → LLM → Tool Selection → Tool Execution → Tool Result → LLM → Final Response**

The available tools are:

- 🧮 Calculator

## Agent Frameworks

Several frameworks can be used to build more advanced AI agents.

Examples include:

- LangChain

orchestrating, and managing more complex agent workflows.
